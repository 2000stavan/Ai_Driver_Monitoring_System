# IDD downloads → Google Drive (Colab)

Downloads **IDD Detection (22.8 GB)** and **IDD Segmentation Part I (18.5 GB)** from the IIIT Hyderabad server straight into your Drive (`My Drive/DMS/raw/idd/`). Nothing goes through your laptop.

- The IIIT server sends ~1-2 MB/s, so each file can take **3-6 hours**. `wget -c` writes directly into Drive, so if Colab disconnects, **rerun all cells and it resumes**.
- Keep this tab open. Run one dataset at a time (cell 3, then cell 4).

**Before running:** on https://idd.insaan.iiit.ac.in/ click *download* for each dataset, then right-click the generated link → *Copy link address*, and paste it into cell 2.

```
# 1. Mount Drive
from google.colab import drive
drive.mount('/content/drive')
!mkdir -p /content/drive/MyDrive/DMS/raw/idd
!df -h /content | tail -1
```

```
# 2. Paste the download links generated on the IDD website (keep the quotes)
IDD_DETECTION_URL    = 'PASTE_IDD_DETECTION_LINK_HERE'
IDD_SEGMENTATION_URL = 'PASTE_IDD_SEGMENTATION_PART1_LINK_HERE'
OUT = '/content/drive/MyDrive/DMS/raw/idd'

# Quick check that a link gives the file (not a login page): shows status, size and type
for name, url in [('detection', IDD_DETECTION_URL), ('segmentation', IDD_SEGMENTATION_URL)]:
    if url.startswith('PASTE'):
        print(name, ': no link yet'); continue
    print('==', name)
    !curl -sIL --max-time 30 "$url" | grep -iE "^HTTP|content-length|content-type|content-disposition" | tail -4
```

```
# 3. IDD Detection (22.8 GB) -> Drive, resumable
!wget -c --progress=bar:force:noscroll -O "$OUT/idd_detection.tar.gz" "$IDD_DETECTION_URL"
!ls -lh "$OUT"
```

```
# 4. IDD Segmentation Part I (18.5 GB) -> Drive, resumable
!wget -c --progress=bar:force:noscroll -O "$OUT/idd_segmentation_part1.tar.gz" "$IDD_SEGMENTATION_URL"
!ls -lh "$OUT"
```

```
# 5. Check the downloads are complete archives (lists the first entries without extracting)
import subprocess, os
for f in ['idd_detection.tar.gz', 'idd_segmentation_part1.tar.gz']:
    p = os.path.join(OUT, f)
    if not os.path.exists(p):
        print(f, ': not downloaded yet'); continue
    print(f'== {f}: {os.path.getsize(p)/1e9:.2f} GB')
    r = subprocess.run(f'tar -tzf "{p}" | head -8', shell=True, capture_output=True, text=True)
    print(r.stdout or r.stderr)
```

**If cell 2 shows `content-type: text/html`** (a web page, not a file) the link only works inside your logged-in browser. Then download in the browser instead, or tell Claude and we'll find another way.
**Note:** if the file name the site gives ends in `.zip` or `.tar` instead of `.tar.gz`, change the file names in cells 3-5 to match.
